## 2) Fachbegriffe
- **imperativ**: Programmierung als Abfolge von Zustandsänderungen (Variablen werden verändert, typische Schleifen/Anweisungen).
- **funktional**: Programmierung über Funktionen/Ausdrücke; Ziel: wenig/keine Seiteneffekte, klarer Datenfluss Input→Output.
- **seiteneffekt**: Eine Funktion verändert Zustand ausserhalb der Rückgabe (z. B. globale Variable/Liste, IO).
- **mutability**: Mutable Objekte (list/dict) sind veränderlich; immutable (int/str/tuple) nicht.



- **call by value**: wert kopierten, variable bleibt unverändert

- **call by reference**: speicher-adresse zuweisen, wert der variablen ändert

Call by value besser für funktionale Programmierung, weil wir unveränderte Variablen wollen.

- **pure/impure functions**: keine side-effects, gleiches Ergebnis braucht gleichen input (referenzielle Transparenz!)

- **Avoiding shared state**: wir wollen call by value
Side-effects: ungewollter zustand der nicht unter Kontrolle ist

- **Funktionale Programmierung** - deklarativ
deklarativ: Haskell, Lisp, R etc.
imperativ: C#, Java, C++ etc.


"Functional programming is the process by building software by composing pure functions, avoiding shared state, mutable data, and side-effects. Functional programming is declarative rather than imperative, and application state flows through pure functions."

### `Aufgabe1.1-Pythagoras_funktional.py`

```python
# Funktionale Programmierung des Pythagoras
# Problembeschreibung:
# Berechnung der Hypotenuse als Wurzel aus der Summe der beiden quadrierten Katheten

import math

def quadrat(kat):
    return kat*kat

def wurzel(base):
    return base**0.5

def summe(sum1, sum2):
    return sum1 + sum2

def hypotenuse(kat1, kat2):
    hypotenuse = math.sqrt(kat1*kat1 + kat2*kat2)
    return hypotenuse

def hypotenuse2(k1, k2):
    return wurzel(summe(quadrat(k1), quadrat(k2)))


print(hypotenuse2(3, 4))

print(wurzel(summe(quadrat(3), quadrat(4))))
```
