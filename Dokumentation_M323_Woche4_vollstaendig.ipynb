## Rekursion Grundidee

rekursion bedeutet, dass sich eine funktion selbst aufruft. dabei wird ein problem so lange vereinfacht, bis ein einfaches abbruchkriterium erreicht ist. typische beispiele sind summen, fakultät oder die fibonacci reihe.


## Beispiel: Summe einer Zahlenreihe

anstatt eine schleife zu verwenden, kann die summe von 0 bis n rekursiv berechnet werden. die funktion ruft sich selbst mit einem kleineren parameter auf.


```python
def sum_rec(n):
    return n + sum_rec(n-1) if n > 0 else 0

sum_rec(10)
```

## Vorteile und Probleme der Rekursion

rekursion passt gut zur funktionalen programmierung, da sie oft ohne seiteneffekte auskommt. allerdings kann sie auch zu problemen führen, zum beispiel durch ineffizienz oder stack overflow, wenn das abbruchkriterium fehlt oder falsch definiert ist.


## Rekursive Algorithmen: suchen

ein klassisches beispiel ist die binäre suche. dabei wird eine sortierte liste immer wieder halbiert, bis das gesuchte element gefunden wird oder feststeht, dass es nicht vorhanden ist.


```python
def bin_search(liste, element, offset=0):
    if not liste:
        return None
    mitte = len(liste) // 2
    if liste[mitte] == element:
        return offset + mitte
    if element < liste[mitte]:
        return bin_search(liste[:mitte], element, offset)
    return bin_search(liste[mitte+1:], element, offset + mitte + 1)

bin_search([1,2,3,4,5,6,7,8,9], 6)
```

## Rekursive Grafiken und Fraktale

mit rekursion lassen sich auch grafische muster erzeugen. durch wiederholtes zeichnen leicht veränderter elemente entstehen raktale strukturen, zum beispiel bäume.


```python
# vereinfachtes beispiel für rekursiven baum (turtle)
import turtle

def draw_tree(depth, length):
    if depth == 0:
        return
    turtle.forward(length)
    turtle.left(30)
    draw_tree(depth-1, length*0.7)
    turtle.right(60)
    draw_tree(depth-1, length*0.7)
    turtle.left(30)
    turtle.backward(length)
```
