## Funktionale Grundlagen

funktionale programmierung setzt auf pure functions, vermeidet seiteneffekte und arbeitet deklarativ. zustände werden nicht verändert, sondern neue werte zurückgegeben.


## Verbesserung einer impuren Lösung

ausgehend von einer impure lösung mit globalen variablen wurde der code so umgebaut, dass keine globalen zustände mehr verwendet werden.


```python
def update_score(user, new_score):
    user_copy = user.copy()
    user_copy['score'] += new_score
    return user_copy
```

## Abstraktion und funktionen höherer Ordnung

abstraktion erlaubt es, code zu verallgemeinern. funktionen können als parameter übergeben oder als rückgabewert genutzt werden.


```python
def mal(x):
    def mal_x(y):
        return x * y
    return mal_x

mal_zwei = mal(2)
mal_zwei(5)
```

## map und reduce

map() wendet eine funktion auf alle elemente einer liste an. reduce() fasst eine liste schrittweise zu einem einzelnen wert zusammen.


```python
from functools import reduce

def mult(x, y):
    return x * y

reduce(mult, range(1, 6))
```
