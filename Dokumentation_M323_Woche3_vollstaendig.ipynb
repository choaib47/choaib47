## Datenstrukturen

neben listen wurden weitere wichtige iterierbare datenstrukturen behandelt. dazu gehören sets, tupel und dictionaries. diese datenstrukturen lassen sich gut mit funktionen höherer ordnung verarbeiten.


```python
# beispiele datenstrukturen
a = {1, 2, 3, 4}
b = {4, 5, 6, 7}
a.union(b)
```

## zip als Funktion höherer Ordnung

die funktion zip() kombiniert mehrere iterierbare datenstrukturen zu einer liste von tupeln. jedes tupel enthält jeweils ein element aus jeder ursprungsliste.


```python
l1 = [1, 2, 3]
l2 = [4, 5, 6]
list(zip(l1, l2))
```

## Filter als Funktion höherer Ordnung

filter() selektiert elemente aus einer datenstruktur anhand einer bedingung, die als funktion übergeben wird.


```python
alter = [12, 3, 6, 17, 34, 45, 56, 90]
kids = list(filter(lambda a: a < 18, alter))
kids
```

## Lambda Funktionen

lambda funktionen sind anonyme funktionen, die direkt an der verwendungsstelle definiert werden. sie eignen sich für kurze berechnungen und werden häufig zusammen mit map(), filter() oder sorted() eingesetzt.


```python
# lambda beispiel
mult = lambda x, y: x * y
mult(5, 7)
```
