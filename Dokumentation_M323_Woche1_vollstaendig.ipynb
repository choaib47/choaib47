## Pure functions und Side effects

pure functions liefern bei gleichem input immer den gleichen output und verursachen keine seiteneffekte. impure functions hingegen greifen auf globale variablen zu oder verändern zustande ausserhalb der funktion.


```python
# pure function
def mult(a, b):
    return a * b

# impure function
x = 2
def mult_impure(b):
    global x
    x = x * b
    return x
```

## Beispiel: funktional vs imperativ

berechnung der summe von 0 bis n; funktional wird das problem rekursiv gelöst, ohne zustände zu verändern.


```python
# funktionale loesung
def summe(n):
    if n == 0:
        return 0
    return n + summe(n-1)
```

## Übung: Pythagoras

satz des pythagoras sowohl imperativ als auch funktional umsetzen; dabei zeigte sich, dass die funktionale variante klarer und besser testbar ist.


```python
# funktionale pythagoras loesung
import math

def pythagoras(a, b):
    return math.sqrt(a*a + b*b)
```
