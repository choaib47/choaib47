## List comprehensions

list comprehensions bieten eine kompakte und deklarative möglichkeit, listen zu erzeugen oder zu transformieren. sie können oft kombinationen aus map(), filter() und lambda-funktionen ersetzen.


```python
# beispiel list comprehension
celsius = [19.2, 26.5, 17.3, 27.8]
absolut = [abs(x) for x in celsius]
absolut
```

## Generator comprehensions

generator comprehensions funktionieren ähnlich wie list comprehensions, erzeugen die werte jedoch erst bei bedarf. dadurch sind sie speichereffizient und eignen sich gut für grosse datenmengen oder zahlenreihen.


```python
# generator comprehension
generator = (i * 3 for i in range(1, 10))
list(generator)
```

## Set comprehensions

set comprehensions erzeugen mengen statt listen. sie stellen sicher, dass jedes element nur einmal vorkommt und eignen sich gut für eindeutige wertebereiche.


```python
# set comprehension
menge = {i * 3 for i in range(10)}
menge
```

## Funktionskomposition

bei der funktionskomposition werden mehrere funktionen miteinander verschachtelt. das ergebnis einer funktion dient dabei als eingabe für die nächste funktion.


```python
def f(x):
    return 2 * x

def g(x):
    return x * x + 1

f(g(2))
```

## Currying

currying ist ein funktionales design-pattern. dabei wird eine funktion mit mehreren parametern in eine folge von funktionen mit jeweils genau einem parameter zerlegt. benannt ist dieses konzept nach haskell curry.


```python
from functools import partial

def mult(x, y, z):
    return x * y * z

mult_10 = partial(mult, 10)
mult_10_20 = partial(mult_10, 20)
mult_10_20(30)
```
